## Q1: NumPy 1D Array — Basic Operations

Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:

**a)** Addition of 2 to all the elements  
**b)** Multiplication of 3 with all the elements  
**c)** Divide every element by 2 

```python
arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)
add_result = arr + 2
print("After adding 2:", add_result)
mul_result = arr * 3
print("After multiplying by 3:", mul_result)
divide_by_2 = arr / 2
print("After division: ", divide_by_2)
```

## Q2: Questions on Basic NumPy Array

**a)** Reverse the NumPy array:  
`arr = np.array([1, 2, 3, 6, 4, 5])`

**b)** Find the most frequent value and their indice(s) in the following arrays:  
&nbsp;&nbsp;&nbsp;&nbsp;i. `x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])`  
&nbsp;&nbsp;&nbsp;&nbsp;ii. `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Reversed array:", reversed_arr)
def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    most_frequent_values = values[counts == max_count]
    indices = np.where(np.isin(arr, most_frequent_values))[0]
    return most_frequent_values, indices

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

x_val, x_idx = most_frequent(x)
print("x -> Most frequent value(s):", x_val, "at indices:", x_idx)

y_val, y_idx = most_frequent(y)
print("y -> Most frequent value(s):", y_val, "at indices:", y_idx)
```

## Q3: Accessing Elements in a 2-D Array

For the given 2-D array:  
`arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])`

Access elements using row and column indices as follows:

**a)** Access 1st row, 2nd column  
**b)** Access 3rd row, 1st column

## Q.4  Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
Waris = np.linspace(10, 100, 25)
print("Array:", Waris)
# Dimensions, shape, size, dtype, bytes
print("\nDimensions (ndim):",Waris.ndim)
print("Shape:", Waris.shape)
print("Total elements (size):", Waris.size)
print("Data type (dtype):", Waris.dtype)
print("Total bytes consumed (nbytes):", Waris.nbytes)
# Transpose using reshape()
Waris_reshaped = Waris.reshape(25, 1)
print("\nTransposed using reshape(25,1):\n",Waris_reshaped)
# Transpose using T attribute
Waris_T = Waris.T
print("\nTransposed using .T attribute:\n",Waris_T)
print("Shape after .T:", Waris_T.shape)
```

## Q5. Create a 2-D Array of three rows and four columns, named ucs420_<<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<<your_name>>.

```python
ucs420_Waris = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])
print("ucs420_Waris:\n", ucs420_Waris)
print("\nMean:", np.mean(ucs420_Waris))
print("Median:", np.median(ucs420_Waris))
print("Max:", np.max(ucs420_Waris))
print("Min:", np.min(ucs420_Waris))
print("Unique elements:", np.unique(ucs420_Waris))
reshaped_ucs420_Waris = ucs420_Waris.reshape(4, 3)
print("\nreshaped_ucs420_Waris:\n", reshaped_ucs420_Waris)
resized_ucs420_Waris = np.resize(ucs420_Waris, (2, 3))
print("\nresized_ucs420_Waris:\n", resized_ucs420_Waris)
```
